##Passo 1: Criar o campo de dados

```python

catalog = "catalogo_de_filmes"
bronze_schema_name = "bronze"
silver_schema_name = "silver"
gold_schema_name = "gold"

bronze_schema = f"{catalog}.{bronze_schema_name}"
silver_schema = f"{catalog}.{silver_schema_name}"
gold_schema = f"{catalog}.{gold_schema_name}"

landing_path = f"/Volumes/{catalog}/{bronze_schema_name}/Landing"

print(f"catalog: {catalog}")
print(f"bronze_schema: {bronze_schema}")
print(f"silver_schema: {silver_schema}")
print(f"gold_schema: {gold_schema}")
print(f"landing_path: {landing_path}")
```

#Mapeamento dos arquivos

```python
tb_movies_info = f"{landing_path}/movies_info_TMDB_IMDB.csv"
tb_movies_financials = f"{landing_path}/movies_financials_IMDB_TMDB.csv"
tb_movies_metrics = f"{landing_path}/movies_metrics_IMDB_TMDB.csv"
tb_credits_and_tags = f"{landing_path}/credits_and_tags_IMDB_TMDB.csv"
tb_movies_reviews = f"{landing_path}/movies_reviews.csv"

df_movies_info_raw = spark.read.csv(tb_movies_info, header=True, inferSchema=True)
df_movies_financials_raw = spark.read.csv(tb_movies_financials, header=True, inferSchema=True)
df_movies_metrics_raw = spark.read.csv(tb_movies_metrics, header=True, inferSchema=True)
df_credits_and_tags_raw = spark.read.csv(tb_credits_and_tags, header=True, inferSchema=True)
df_movies_reviews_raw = spark.read.csv(tb_movies_reviews, header=True, inferSchema=True)
display(df_movies_info_raw)
display(df_movies_financials_raw)
display(df_movies_metrics_raw)
display(df_credits_and_tags_raw)
display(df_movies_reviews_raw)
```
